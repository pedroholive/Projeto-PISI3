# Análise Exploratória de Dados — HealthSync

## 1. Contextualização da base

Esta análise exploratória utiliza dados do **NHANES 2021–2023**, selecionados com o objetivo de investigar indicadores relacionados à saúde cardiometabólica.

A análise está alinhada ao **HealthSync**, sistema voltado ao acompanhamento de indicadores como glicemia e pressão arterial, além de informações relacionadas ao perfil e estilo de vida dos usuários.

O conjunto principal utilizado nesta análise contém adultos com 20 anos ou mais que possuem informações disponíveis de IMC, pressão arterial, hemoglobina glicada (HbA1c) e colesterol total.

As principais dimensões analisadas incluem:

- características demográficas, como idade e sexo;
- índice de massa corporal (IMC);
- pressão arterial sistólica e diastólica;
- hemoglobina glicada (HbA1c);
- colesterol total;
- glicose em jejum, disponível em uma subamostra;
- atividade física moderada e vigorosa;
- comportamento sedentário;
- histórico informado de diabetes, hipertensão e colesterol elevado.

### Objetivo da análise exploratória

A EDA tem como objetivo compreender a estrutura e a qualidade dos dados antes da etapa de Machine Learning. Para isso, são analisadas as distribuições das principais variáveis, a presença de valores ausentes e extremos e as relações entre diferentes indicadores cardiometabólicos.

Essa etapa também permite identificar características dos dados que podem influenciar decisões posteriores de pré-processamento, seleção de variáveis e definição dos modelos.

Por se tratar de um estudo transversal, as relações observadas representam associações entre as características dos participantes avaliados e não devem ser interpretadas automaticamente como relações causais ou como mudanças individuais ao longo do tempo.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
```

```python
# ============================================================
# CONFIGURAÇÃO VISUAL — HEALTHSYNC
# ============================================================

sns.set_theme(
    style="whitegrid",
    context="notebook"
)

# Identidade visual do HealthSync
COR_PRINCIPAL = "#0E766D"
COR_SECUNDARIA = "#10B981"
COR_TERCIARIA = "#14B8A6"

COR_DESTAQUE = "#F28C82"
COR_NEUTRA = "#94A3B8"

# Cores consistentes para respostas binárias
COR_SIM = COR_PRINCIPAL
COR_NAO = COR_NEUTRA

plt.rcParams["figure.figsize"] = (10, 6)

plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 11

plt.rcParams["xtick.labelsize"] = 10
plt.rcParams["ytick.labelsize"] = 10

plt.rcParams["figure.dpi"] = 120
```

```python
CAMINHO_CORE = "../../dataset/healthsync_nhanes_2021_2023_core.csv"

df = pd.read_csv(CAMINHO_CORE)

print("Dataset carregado com sucesso.")
print(f"Participantes: {df.shape[0]}")
print(f"Variáveis: {df.shape[1]}")

display(df.head())
```

### Base utilizada nas análises

O conjunto CORE contém **5.283 participantes e 23 variáveis** e será utilizado como base principal da análise exploratória.

A visualização das primeiras observações permite verificar a estrutura do conjunto e confirmar a presença das variáveis demográficas, clínicas, comportamentais e de histórico de saúde selecionadas para o estudo.

A glicose em jejum possui disponibilidade apenas em uma subamostra dos participantes. Por esse motivo, análises específicas envolvendo essa variável serão realizadas utilizando o conjunto FASTING, evitando reduzir desnecessariamente a amostra utilizada nas demais análises.

## 3. Qualidade e estrutura dos dados

### Objetivo

Antes de analisar as características clínicas e comportamentais dos participantes, é necessário avaliar a estrutura e a qualidade do conjunto de dados.

Nesta etapa são verificadas:

- as dimensões da base;
- os tipos das variáveis;
- a existência de participantes ou registros duplicados;
- a presença e o padrão de valores ausentes;
- as principais estatísticas descritivas das variáveis numéricas.

Essas verificações são importantes para identificar possíveis problemas que possam interferir nas análises posteriores e na etapa de Machine Learning.

### 3.1 Dimensão

```python
print(f"Número de participantes: {df.shape[0]}")
print(f"Número de variáveis: {df.shape[1]}")
```

### 3.2 Tipos das variáveis

```python
tipos = pd.DataFrame({
    "tipo": df.dtypes
})

display(tipos)
```

A inspeção dos tipos confirma que a maior parte das variáveis clínicas e comportamentais foi carregada em formato numérico, enquanto variáveis categóricas, como sexo e diabetes informado, são representadas como texto.

Os campos `peso_estatistico_mec` e `peso_estatistico_entrevista`, entretanto, foram carregados como texto (`str`). Esses campos correspondem a pesos estatísticos do desenho amostral e não a medidas clínicas ou de massa corporal.

Como esses pesos não são utilizados nas análises descritivas não ponderadas apresentadas neste notebook, o problema de formatação não interfere nos resultados atuais. Entretanto, sua representação numérica deverá ser corrigida antes de qualquer análise que utilize ponderação amostral.

### 3.3 Duplicatas

O identificador `id_participante` corresponde ao `SEQN`, utilizado pelo NHANES para identificar cada participante.

Como cada linha da base analítica deve representar um único participante, são verificadas duas situações:

- repetição do identificador do participante;
- repetição completa de registros.

A presença de duplicatas poderia alterar contagens, proporções e estatísticas calculadas posteriormente.

```python
duplicados_id = df["id_participante"].duplicated().sum()

duplicados_linha = df.duplicated().sum()

print(f"IDs duplicados: {duplicados_id}")
print(f"Linhas completamente duplicadas: {duplicados_linha}")
```

Não foram encontrados identificadores duplicados nem linhas completamente repetidas.

Dessa forma, cada um dos **5.283 registros** da base CORE representa uma observação única dentro do conjunto utilizado na análise.

### 3.4 Valores ausentes

A presença de valores ausentes é avaliada antes das análises individuais para identificar quais variáveis possuem menor cobertura na amostra.

Entretanto, nem todo valor ausente possui o mesmo significado. No NHANES, determinadas variáveis são coletadas apenas em subamostras ou dependem das respostas fornecidas anteriormente no questionário. Portanto, percentuais elevados de ausência não devem ser interpretados automaticamente como falhas na coleta ou problemas de qualidade dos dados.

```python
missing = pd.DataFrame({
    "ausentes": df.isna().sum(),
    "percentual": (
        df.isna()
        .mean()
        .mul(100)
        .round(2)
    )
})

missing = (
    missing[
        missing["ausentes"] > 0
    ]
    .sort_values(
        "percentual",
        ascending=False
    )
)

display(missing)
```

```python
missing_plot = (
    missing
    .reset_index()
    .rename(columns={
        "index": "variavel"
    })
)

nomes_variaveis = {
    "hipertensao_2consultas": "Hipertensão em 2+ consultas",
    "medicamento_pressao": "Medicamento para pressão",
    "glicose_jejum": "Glicose em jejum",
    "minutos_sedentarios_dia": "Tempo sedentário",
    "atv_moderada_min_semana": "Atividade moderada",
    "colesterol_alto_informado": "Colesterol alto informado",
    "atv_vigorosa_min_semana": "Atividade vigorosa",
    "hipertensao_informada": "Hipertensão informada"
}

missing_plot["variavel"] = (
    missing_plot["variavel"]
    .replace(nomes_variaveis)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=missing_plot,
    y="variavel",
    x="percentual",
    color=COR_PRINCIPAL
)

plt.title(
    "Percentual de valores ausentes por variável",
    pad=15,
    weight="bold"
)

plt.xlabel("Valores ausentes (%)")
plt.ylabel("Variável")

sns.despine()

plt.tight_layout()
plt.show()
```

A maior parte das variáveis principais apresenta baixa proporção de valores ausentes.

As maiores proporções aparecem em `hipertensao_2consultas` (62,90%), `medicamento_pressao` (62,63%) e `glicose_jejum` (44,07%). Entretanto, esses casos possuem características específicas do processo de coleta.

As perguntas relacionadas a consultas e medicamentos fazem parte de fluxos condicionais do questionário, fazendo com que determinados participantes não precisem respondê-las. Dessa forma, parte dos valores ausentes representa **saltos estruturais do questionário**, e não simplesmente dados perdidos.

A glicose em jejum, por sua vez, está disponível apenas para uma subamostra específica. Por esse motivo, sua ausência na base CORE não será utilizada como critério para exclusão de participantes.

Nas demais variáveis apresentadas, a proporção de valores ausentes permaneceu abaixo de 1%, indicando alta disponibilidade dessas informações na amostra utilizada.

### 3.5 Estatísticas descritivas

Após a avaliação estrutural da base, são calculadas estatísticas descritivas das principais variáveis numéricas de interesse.

São apresentados número de observações válidas, média, desvio-padrão, mínimo, quartis, mediana e máximo. Essas medidas permitem obter uma visão inicial da tendência central, dispersão e amplitude dos dados antes da análise gráfica individual de cada variável.

Nesta etapa são consideradas principalmente variáveis demográficas, clínicas e comportamentais. Os pesos estatísticos e identificadores não são incluídos por não representarem características clínicas dos participantes.

```python
variaveis_numericas = [
    "idade",
    "imc",
    "pressao_sistolica",
    "pressao_diastolica",
    "hba1c",
    "glicose_jejum",
    "colesterol_total",
    "atv_moderada_min_semana",
    "atv_vigorosa_min_semana",
    "minutos_sedentarios_dia"
]

resumo = (
    df[variaveis_numericas]
    .describe()
    .T
    .round(2)
)

display(resumo)
```

**Observação inicial**

As estatísticas mostram que as variáveis possuem escalas e distribuições bastante diferentes. Também são observados valores máximos consideravelmente afastados dos quartis em algumas medidas, principalmente nas variáveis de atividade física.

Neste momento, esses valores não serão removidos automaticamente. A presença de uma observação distante da região central da distribuição não é suficiente para classificá-la como erro.

Cada grupo de variáveis será analisado com maior detalhe nas próximas seções por meio de estatísticas específicas e visualizações adequadas.

## 4. Perfil da amostra

### Objetivo

Esta seção descreve características básicas dos participantes da base CORE, considerando idade, sexo e IMC.

Compreender a composição da amostra é importante porque diferenças na distribuição dessas características podem influenciar a interpretação dos indicadores cardiometabólicos analisados posteriormente.

As análises apresentadas descrevem os participantes disponíveis na base utilizada e não devem ser interpretadas automaticamente como estimativas da população geral.

### 4.1 Idade

**Pergunta da análise:** como as idades estão distribuídas entre os participantes da amostra?

A idade é analisada inicialmente por meio de um histograma, permitindo visualizar a concentração e a distribuição dos participantes ao longo das diferentes idades.

Como os dados são transversais, essa visualização representa a distribuição etária da amostra e não uma evolução dos mesmos participantes ao longo do tempo.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="idade",
    bins=20,
    color=COR_PRINCIPAL,
    edgecolor="white"
)

plt.title(
    "Distribuição da idade dos participantes",
    pad=15,
    weight="bold"
)

plt.xlabel("Idade (anos)")
plt.ylabel("Número de participantes")

sns.despine()

plt.tight_layout()
plt.show()
```

```python
df["idade"].describe().round(2)
```

A amostra apresenta participantes entre **20 e 80 anos**, com média de **54,02 anos** e mediana de **58 anos**. Metade dos participantes encontra-se aproximadamente entre **40 e 68 anos**, intervalo definido pelo primeiro e terceiro quartis.

Observa-se maior concentração de participantes nas faixas etárias mais elevadas da amostra. Essa distribuição deve ser considerada nas análises posteriores que envolvam idade e indicadores cardiometabólicos.

Por se tratar de dados transversais, diferenças observadas entre participantes de diferentes idades não representam o acompanhamento do envelhecimento dos mesmos indivíduos.

### 4.2 Sexo

**Pergunta da análise:** como os participantes estão distribuídos segundo o sexo registrado na base?

Por se tratar de uma variável categórica, a distribuição é apresentada por meio de barras horizontais, permitindo comparar diretamente a participação relativa de cada categoria.

```python
sexo = (
    df["sexo"]
    .value_counts()
    .rename_axis("sexo")
    .reset_index(name="quantidade")
)

sexo["percentual"] = (
    sexo["quantidade"]
    / sexo["quantidade"].sum()
    * 100
)

display(sexo)
```

```python
plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=sexo,
    y="sexo",
    x="percentual",
    color=COR_PRINCIPAL
)

plt.title(
    "Distribuição dos participantes por sexo",
    pad=15,
    weight="bold"
)

plt.xlabel("Participantes (%)")
plt.ylabel("Sexo")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=4
    )

sns.despine()

plt.tight_layout()
plt.show()
```

A amostra apresenta **54,6% de mulheres** e **45,4% de homens**.

Existe, portanto, uma participação ligeiramente maior de mulheres, embora ambas as categorias estejam amplamente representadas no conjunto analisado.

### 4.3 IMC

**Pergunta da análise:** como os valores de IMC estão distribuídos entre os participantes?

O IMC é uma das principais medidas antropométricas disponíveis na base e será considerado em análises posteriores envolvendo o perfil cardiometabólico.

Para compreender sua distribuição, são utilizados dois gráficos complementares:

- **histograma:** permite observar o formato geral e a concentração dos valores;
- **boxplot:** resume a distribuição por meio da mediana e dos quartis e permite identificar visualmente observações mais afastadas da região central.

A identificação de valores extremos no boxplot não significa, por si só, que essas observações sejam erros.

```python
# ============================================================
# DISTRIBUIÇÃO DO IMC
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="imc",
    bins=30,
    kde=True,
    color=COR_PRINCIPAL,
    edgecolor="white"
)

plt.title(
    "Distribuição do IMC dos participantes",
    pad=15,
    weight="bold"
)

plt.xlabel("IMC (kg/m²)")
plt.ylabel("Número de participantes")

sns.despine()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# BOXPLOT DO IMC
# ============================================================

plt.figure(figsize=(10, 3.5))

sns.boxplot(
    data=df,
    x="imc",
    color=COR_SECUNDARIA
)

plt.title(
    "Distribuição e valores extremos do IMC",
    pad=15,
    weight="bold"
)

plt.xlabel("IMC (kg/m²)")
plt.ylabel("")

sns.despine(
    left=True
)

plt.tight_layout()
plt.show()
```

**Interpretação**

O IMC apresentou média de **29,72 kg/m²** e mediana de **28,50 kg/m²**, com metade dos participantes apresentando valores entre aproximadamente **24,7 e 33,5 kg/m²**.

A distribuição apresenta assimetria à direita, com maior concentração dos participantes em valores intermediários e menor frequência à medida que o IMC aumenta.

O boxplot também evidencia observações mais afastadas da região central da distribuição. Esses valores foram preservados, pois sua posição no gráfico não é suficiente para classificá-los como erros de registro.

### 4.4 Faixas de IMC

Além da análise do IMC como variável contínua, são utilizadas faixas apenas para facilitar a descrição da composição da amostra.


```python
faixas_imc = [
    0,
    18.5,
    25,
    30,
    35,
    40,
    np.inf
]

labels_imc = [
    "Abaixo de 18,5",
    "18,5–24,9",
    "25,0–29,9",
    "30,0–34,9",
    "35,0–39,9",
    "40 ou mais"
]

df["faixa_imc"] = pd.cut(
    df["imc"],
    bins=faixas_imc,
    labels=labels_imc,
    right=False
)

distribuicao_imc = (
    df["faixa_imc"]
    .value_counts(sort=False)
    .rename_axis("faixa_imc")
    .reset_index(name="quantidade")
)

distribuicao_imc["percentual"] = (
    distribuicao_imc["quantidade"]
    / distribuicao_imc["quantidade"].sum()
    * 100
)

display(distribuicao_imc)
```

```python
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=distribuicao_imc,
    y="faixa_imc",
    x="percentual",
    color=COR_PRINCIPAL
)

plt.title(
    "Distribuição dos participantes por faixa de IMC",
    pad=15,
    weight="bold"
)

plt.xlabel("Participantes (%)")
plt.ylabel("Faixa de IMC (kg/m²)")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=4
    )

sns.despine()

plt.tight_layout()
plt.show()
```

A faixa com maior participação na amostra foi de **25,0 a 29,9 kg/m²**, correspondendo a aproximadamente **32,6%** dos participantes.

Em seguida aparecem as faixas de **18,5 a 24,9 kg/m² (25,4%)** e **30,0 a 34,9 kg/m² (21,1%)**. As faixas de 35,0 a 39,9 kg/m² e de 40 kg/m² ou mais corresponderam a aproximadamente 11,0% e 8,8% da amostra, respectivamente.

A categorização complementa a análise da distribuição contínua do IMC e facilita a descrição da composição da amostra, sem substituir os valores originais utilizados nas demais análises.

As características observadas nesta seção descrevem a amostra disponível e serão consideradas na interpretação das relações cardiometabólicas investigadas posteriormente.
